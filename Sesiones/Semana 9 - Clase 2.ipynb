## Series estacionarias

* La media es fija sobre el eje horizontal
* La varianza es constante (homocedasticidad)

* **Efecto de choque**: shock / perturbación $\varepsilon_i$
* **Raíz unitaria**: Dependencia a valores pasados
* **Proceso autoregresivo**: Valor actual como combinación lineal de los valores pasados

### Prueba de Dickey-Fuller Aumentada (ADF)

* $H_0$ - Serie no estacionaria
* $H_1$ - Serie es estacionaria

* $p_{valor} > 0.5$ - No se rechaza $H_0$ y no es estacionaria
* $p_{valor} \le 0.5$ - Se rechaza $H_0$ y es estacionaria

```python
from statsmodels.tsa.stattools import adfuller
```

```python
resultado = adfuller(ford["Close"])

reporte = pandas.DataFrame({
    "Estadístico ADF": [resultado[0]],
    "p-value": [resultado[1]],
    "Lags usados": [resultado[2]],
    "Número de observaciones": [resultado[3]],
    "Valor crítico 1%": [resultado[4]["1%"]],
    "Valor crítico 5%": [resultado[4]["5%"]],
    "Valor crítico 10%": [resultado[4]["10%"]],
}, index=["Valor"]).T

reporte
```

## Persistencia

### Exponente de Hurst

* $H = 0.5$ - Paseo aleatorio
* $H < 0.5$ - Antipersistente
* $H > 0.5$ - Persistente
* $H = 1$ - Persistencia perfecta

Serie persistente

```python
n = 500

serie_p = [0]

for t in range(1, n):
    x = 0.9 * serie_p[-1] + numpy.random.normal(0, 1)
    serie_p.append(x)

serie_p = pandas.Series(serie_p)

serie_a = [0]

for t in range(1, n):
    x = -0.9 * serie_a[-1] + numpy.random.normal(0, 1)
    serie_a.append(x)

serie_a = pandas.Series(serie_a)
```

```python
figure, axis = pyplot.subplots(2)

serie_p.plot(ax=axis[0])
serie_a.plot(ax=axis[1], color="orange")
```

### Hurst

`compute_Hc(ts, kind="change", simplified=True)`

* `ts` - Serie temporal
* `kind` - `change` cambios o diferencias o `price` serie acumulada
* `simplified` - Si se usa rango 

```python
from hurst import compute_Hc
```

```python
H_p = compute_Hc(serie_p, kind="change")
H_p[0]
```

```python
H_a = compute_Hc(serie_a, kind="change")
H_a[0]
```
